````python
# Import required libraries
from faker import Faker
import pandas as pd
from datetime import datetime, timedelta

# Initialize Faker
fake = Faker()

# Custom provider for healthcare specific data
class HealthcareProvider:
    def mrn(self):
        return fake.unique.random_number(digits=8)
    
    def icd10(self):
        categories = ['A', 'B', 'C', 'D', 'E', 'F', 'G']
        return f"{fake.random_element(categories)}{fake.random_int(10, 99)}.{fake.random_int(0, 9)}"

# Add provider to faker
fake.add_provider(HealthcareProvider)

# Generate sample patient data
def generate_patient_data(num_records=10):
    data = []
    for _ in range(num_records):
        data.append({
            'mrn': fake.mrn(),
            'first_name': fake.first_name(),
            'last_name': fake.last_name(),
            'dob': fake.date_of_birth(minimum_age=18, maximum_age=90),
            'diagnosis': fake.icd10(),
            'admission_date': fake.date_this_year()
        })
    return pd.DataFrame(data)

# Generate and display sample data
df = generate_patient_data(5)
df
````
